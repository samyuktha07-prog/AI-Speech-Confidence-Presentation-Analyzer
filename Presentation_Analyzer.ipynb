```python
import subprocess
import time
import re

# Stop previous processes
subprocess.run("pkill -f streamlit", shell=True)
subprocess.run("pkill -f cloudflared", shell=True)

# Start Streamlit
streamlit_process = subprocess.Popen(
    [
        "streamlit",
        "run",
        "/content/speaksense_ai.py",
        "--server.address=0.0.0.0",
        "--server.port=8501",
        "--server.headless=true",
        "--browser.gatherUsageStats=false"
    ],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

time.sleep(5)

# Start Cloudflare temporary tunnel
tunnel_process = subprocess.Popen(
    [
        "/content/cloudflared",
        "tunnel",
        "--url",
        "http://localhost:8501",
        "--no-autoupdate"
    ],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

print("=" * 65)
print("🎙️ SPEAKSENSE AI")
print("=" * 65)
print("⏳ Creating secure public tunnel...")
print()

url = None

for _ in range(30):
    line = tunnel_process.stdout.readline()

    if line:
        match = re.search(
            r"https://[a-zA-Z0-9.-]+\.trycloudflare\.com",
            line
        )

        if match:
            url = match.group(0)
            break

    time.sleep(1)

if url:
    print("✅ SpeakSense AI is LIVE!")
    print()
    print("🌐 OPEN THIS LINK:")
    print(url)
    print()
    print("=" * 65)
else:
    print("❌ Could not create the tunnel.")
    print("Run the cell again.")
```
